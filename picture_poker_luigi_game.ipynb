```python
# Cards
import random
from collections import Counter
## Create the deck
# 5 cards of each picture in the deck - 30 cards total
def create_deck(): 
    """
    Function description: This function creates a deck of 30 cards consisting of
        5 cards of each Mario themed card. These cards are then shuffled.
    Inputs/parameters:
        No parameters
    Outputs:
        List of 30 shuffled cards.
    """
    cards = (
        ["Star"] * 5 + 
        ["Mario"] * 5 + 
        ["Luigi"] * 5 + 
        ["Fire Flower"] * 5 + 
        ["Mushroom"] * 5 + 
        ["Cloud"] * 5
        )
    random.shuffle(cards)
    return cards

# Each card is assigned a value for its worth
card_rank = {
    "Star": 6,
    "Mario": 5,
    "Luigi": 4,
    "Fire Flower": 3,
    "Mushroom": 2,
    "Cloud": 1
}

```

```python
# Card hands
# Deal 5 cards to both Luigi and player
def deal_hands(cards):
    """
    Function description: This function deals 5 cards to a player/Luigi.
    Inputs/parameters:
        cards(list): shuffled deck of cards
    Outputs:
        tuple: tuple containing Luigi's and the player's hand, where each
        hand is a list of 5 cards.
    """
    luigi_hand=[]
    player_hand=[]
    # 5 cards, alternate between giving Luigi and player a card
    for i in range(5):
        # append card from deck to hand, pop card from the deck
        luigi_hand.append(cards.pop())
        #print("Luigi's hand: ")
        #print(luigi_hand) # print cards as they are added to luigi's hand

        player_hand.append(cards.pop())
        #print("Player's hand: ")
        #print(player_hand)  # print cards as they are added to player's hand
    return luigi_hand, player_hand
```

```python
# If the player wishes to discard and replace their cards
def discard(player_hand, deck):
    """
    Function description: Allows the player to choose whether to discard and replace cards in their hand.
    Inputs/parameters:
        player_hand: the player's current hand of cards.
        deck: the remaining deck of cards used to replace discarded cards.
    Outputs:
        Returns the updated player hand after replacing selected cards, or the original
        hand if the player chooses not to discard any cards.
    """
    print("Card rankings: Star > Mario > Luigi > Fire flower > Mushroom > Cloud")
    while True:
        discard_confirm = input("Would you like to discard and replace cards? (y/n)\n")
        
        # If yes
        if discard_confirm.lower() == "y":
            # print the player's current hand
            print("Your hand:")

            for i, card in enumerate(player_hand, start=1):
                print(f"{i}. {card}")

            while True:
                discard_choose = input(
                        "----------------------------------------------------\n"
                        "Which cards would you like to discard and replace?\n "
                        "(enter positions like 1 3 5 or press enter to cancel)\n" 
                    )
                # Split input into individual numbers
                positions = discard_choose.split()

                try:
                    indexes = [int(pos) - 1 for pos in positions]   # start from 0
                    
                # If there are letters
                except ValueError:
                    print("Please enter numbers only.")
                    continue
                
                # Press enter to cancel
                if not indexes:
                    return(player_hand)
                
                # Input more than 5 numbers
                if not all(0 <= index <5 for index in indexes):
                    print("Choose numbers 1 to 5 only.")
                    continue
                
                # Same card entered more than once
                if len(indexes) != len(set(indexes)):
                    print("You entered the same card more than once.")
                    continue


                for index in indexes:
                    player_hand[index] = deck.pop()
                
                return player_hand


        
        # If no, just return hand
        elif discard_confirm.lower() == "n":
            return player_hand
        
        # Any other input is not validgit add.
        else:
            print("Not a valid input, please type 'y' or 'n'.\n")
```

```python
# These need to be put on game start
# Start the game off with 10 coins, lose 1 coin for the bet
##coins = 9
#pot = 1
def raise_bet(coins, pot):
    """
    Function description: Allows the player to choose whether to raise the bet and updates the pot and
        the number of coins they have left.
    Inputs/parameters:
        coins: the number of coins the player currently has.
        pot: the current amount of coins in the pot.
    Outputs:
        Returns the updated pot and the updated number of coins after the player
        chooses whether to raise the bet.
    """
    while True:
        raise_confirm = input("Would you like to raise the bet? (y/n):\n")

        if raise_confirm.lower() == "y":
            while True:
                print("-------------------------------\n")
                raise_amount = input("How much would you like to raise?\n")

                try:
                    raise_amount = int(raise_amount)

                # if there are letters
                except ValueError:
                    print("Please enter a whole number only.")
                    continue

                if raise_amount < 0:
                    print("Raise amount must be a positive number.")
                    continue

                if raise_amount == 0:
                    print("You're chickening out? Okay then.")
                    print("Current pot is: ", pot)
                    print("You have", coins, "coins left.")
                    return pot, coins
                
                if raise_amount > coins:
                    print("You're too poor!")
                    continue
                
                pot = pot + raise_amount
                coins = coins - raise_amount

                print("Current pot is: ", pot)
                print("You have", coins, "coins left.")
                return pot, coins

        elif raise_confirm.lower() == "n":
            print("Current pot is: ", pot)
            print("You have", coins, "coins left.")
            return pot, coins
        
        else:
            print("Not a valid input, please type 'y' or 'n'.\n")


```

```python
def hand_eval(hand):
    """
    Function description: Allows player to choose whether to raise the bet and updates the
        pot and number of coins they have left.
    Inputs/parameters:
        coins: number of coins the player currently has.
        pot: number of coins they have left.
    Outputs:
        Returns updated pot and number of coins.
    """
    # count each card type in hand and sort from biggest to smallest
    counts = Counter(hand)
    values = sorted(counts.values(), reverse=True)
    # find the highest value card
    highest_card = max(hand, key=lambda card: card_rank[card])

    # 16 = five-a-kind, 8 = four-of-a-kind, 6 = fullhouse, etc.
    if values == [5]:
        hand_type = 16
    elif values == [4, 1]:
        hand_type = 8
    elif values == [3, 2]:
        hand_type = 6
    elif values == [3, 1, 1]:
        hand_type = 4
    elif values == [2, 2, 1]:
        hand_type = 3
    elif values == [2, 1, 1, 1]:
        hand_type = 2
    else:
        hand_type = 1

    # tuple containing hand type and the highest card rank
    return (hand_type, card_rank[highest_card])
    
    
```

```python
def hand_compare(player_hand, luigi_hand, pot, coins):
    """
    Function description: Compares the player's hand with Luigi's hand, decides the winner,
        and updates the pot and player's coins.
    Input/parameters:
        player_hand: player's final hand of cards.
        luigi_hand: Luigi's final hand of cards.
        pot: the current amount of coins in the pot.
        coins: the player's current number of coins.
    """
    player_score = hand_eval(player_hand)
    luigi_score = hand_eval(luigi_hand)

    print("-----------------------------------\n")

    print("Player hand:", player_hand, "->", player_score)
    print("Luigi hand:", luigi_hand, "->", luigi_score)

    # when player wins round: winnings x pot, add to player coins
    if player_score > luigi_score:
        winnings = pot * player_score[0]
        coins = coins + winnings
        pot = 0
        print("You have won", winnings, "coins!")
        print("You now have", coins, "coins!")
        
        return "Player wins!", pot, coins
    
    # when player loses round: losings x pot, minus from
    elif player_score < luigi_score:
        losings = pot * luigi_score[0]
        coins = max(0, coins - losings)
        pot = 0
        print("You have lost", losings, "coins!")
        print("You now have", coins, "coins!")
        return "Luigi wins!", pot, coins
    else:
        coins = coins + pot
        print("You now have", coins, "coins!")
        return "It's a tie! you get your coins back!", pot, coins
    
```

```python
# player losing hand
luigi_hand = ["Star", "Star", "Star", "Star", "Star"]
player_hand = ["Star", "Mario", "Mario", "Mario", "Cloud"]

# player winning hand
player_hand = ["Star", "Star", "Star", "Star", "Star"]
luigi_hand = ["Star", "Mario", "Mario", "Mario", "Cloud"]

result, pot, coins = hand_compare(player_hand, luigi_hand, pot, coins)
print(result)
```

# Run the Gameplay Loop!!!

```python
# main gameplpay loop
# game stops when you lose
# you are stuck until luigi whoops your butt just like the ds game
def play_game():
    """
    Function description: Run the main gameplay loop, start game with 10 coins
        and continues until player runs out of coins.
    Input/parameters:
        Blank
    Outputs:
        Starts and controls game, printing game progress and ends when
        player has 0 coins left.
    """
    coins = 10

    while coins > 0:
        print("HOW TO PLAY PICTURE POKER LUIGI:\n"
      "Start with 10 coins, draw 5 cards, choose whether to raise or discard, "
      "then compare your hand with Luigi's. \nKeep playing until your coins reach 0.")
        print("\n====================")
        print("Starting new round")
        print("====================")

        pot = 1
        coins -= 1

        print("You put 1 coin into the pot.")
        print("You now have", coins, "coins left.")

        deck = create_deck()
        luigi_hand, player_hand = deal_hands(deck)

        print("\nYour starting hand:", player_hand)

        pot, coins = raise_bet(coins, pot)
        player_hand = discard(player_hand, deck)

        result, pot, coins = hand_compare(player_hand, luigi_hand, pot, coins)
        print(result)

        if coins <= 0:
            break

    print("\nGame over! You ran out of coins.")

play_game()
```
