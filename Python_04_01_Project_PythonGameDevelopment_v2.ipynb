Star > Mario > Luigi > Fire flower > Mushroom > Cloud

5 of each in deck = 30 cards

- create deck
- shuffle deck
- luigi has 5 cards
- you have 5 cards
- discard and replace?

    - yes: input number of coins
    - no: 0
    - valid response?: ask again
- luigi shows his hand
- evaluate hand
    - win: number of coins bet x winning hand value to you
    - lose: number of coins ber x winning hand value to luigi
- coin check:
    - over 0: loop gameplay

```python
# Cards
import random
from collections import Counter
## Create the deck
# 5 cards of each picture in the deck - 30 cards total
def create_deck(): 
    cards = (
        ["Star"] * 5 + 
        ["Mario"] * 5 + 
        ["Luigi"] * 5 + 
        ["Fire Flower"] * 5 + 
        ["Mushroom"] * 5 + 
        ["Cloud"] * 5
        )
    random.shuffle(cards)
    return cards

# Each card is assigned a value for its worth
card_rank = {
    "Star": 6,
    "Mario": 5,
    "Luigi": 4,
    "Fire Flower": 3,
    "Mushroom": 2,
    "Cloud": 1
}

```

```python
## Deck test
deck = create_deck()
## Give list of cards in the deck
## Rerunnning this line of code should reshuffle the deck each time - which it does! >:D
print(deck)

## Check cards in the deck total to 30 - they do total to 30
print("Number of cards: " + str(len(deck)))

## Check there are 5 cards of each picture - which there are
card_count = Counter(create_deck())
print("Number of each picture card:\n" + str(card_count))
```

```python
# Card hands
# Deal 5 cards to both Luigi and player
def deal_hands(cards):
    luigi_hand=[]
    player_hand=[]
    # 5 cards, alternate between giving Luigi and player a card
    for i in range(5):
        # append card from deck to hand, pop card from the deck
        luigi_hand.append(cards.pop())
        #print("Luigi's hand: ")
        #print(luigi_hand) # print cards as they are added to luigi's hand

        player_hand.append(cards.pop())
        #print("Player's hand: ")
        #print(player_hand)  # print cards as they are added to player's hand
    return luigi_hand, player_hand
```

```python
## Dealing hands test
deck = create_deck()   # have to regen the deck each time

luigi_hand, player_hand = deal_hands(deck)

print(luigi_hand)
print(player_hand)
```

```python
# If the player wishes to discard and replace their cards
def discard(player_hand, deck):
    while True:
        discard_confirm = input("Would you like to discard and replace cards? (y/n)\n")
        
        # If yes
        if discard_confirm.lower() == "y":
            # print the player's current hand
            print("Your hand:")

            for i, card in enumerate(player_hand, start=1):
                print(f"{i}. {card}")

            while True:
                discard_choose = input(
                        "Which cards would you like to discard and replace? "
                        "(enter positions like 1 3 5 or press enter to cancel)\n" 
                    )
                # Split input into individual numbers
                positions = discard_choose.split()

                try:
                    indexes = [int(pos) - 1 for pos in positions]   # start from 0
                    
                # If there are letters
                except ValueError:
                    print("Please enter numbers only.")
                    continue
                
                # Press enter to cancel
                if not indexes:
                    return(player_hand)
                
                # Input more than 5 numbers
                if not all(0 <= index <5 for index in indexes):
                    print("Choose numbers 1 to 5 only.")
                    continue
                
                # Same card entered more than once
                if len(indexes) != len(set(indexes)):
                    print("You entered the same card more than once.")
                    continue


                for index in indexes:
                    player_hand[index] = deck.pop()
                
                return player_hand


        
        # If no, just return hand
        elif discard_confirm.lower() == "n":
            return player_hand
        
        # Any other input is not validgit add.
        else:
            print("Not a valid input, please type 'y' or 'n'.\n")
```

```python
# Discard test
discard(player_hand, deck)
```

```python
# These need to be put on game start
# Start the game off with 10 coins, lose 1 coin for the bet
coins = 10
pot = 0
def raise_bet():

    while True:
        raise_confirm = input("Would you like to raise the bet? (y/n):\n")

        if raise_confirm.lower() == "y":
            while True:
                raise_amount = input("How much would you like to raise?\n")

                try:
                    raise_amount = int(raise_amount)

                except ValueError:
                    print("Please enter a whole number only.")
                    continue

                if raise_amount < 0:
                    print("Raise amount must be a positive number.")
                    continue

                if raise_amount == 0:
                    print("You're chickening out? Okay then.")
                    return pot, coins
                
                if raise_amount > coins:
                    print("You're too poor!")
                    continue
                
                pot = pot + raise_amount
                coins = coins - raise_amount

                print("Current pot is: ", pot)
                print("You have", coins, "coins left.")
                return pot, coins

        elif raise_confirm.lower() == "n":
            return pot, coins
        
        else:
            print("Not a valid input, please type 'y' or 'n'.\n")


```

```python
# raise bet check
coins, pot = raise_bet(coins, pot)
```
